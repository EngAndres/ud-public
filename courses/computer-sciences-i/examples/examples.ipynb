### Euclidean GCD Algorithm 

> Use TypeHints to guide developers to use your code

> Use DocStrings to document your functions, classes, and methods

```python

def euclidean_gcd(a: int, b: int) -> int:
    """
    This function implements the euclidean algorithms
    to get the greater common divisor of two numbers.

    Args:
        a(int): First number in the algorithm. It is expected to be the greater one.
        b(int): Second number in the algorithm.

    Returns:
        An integer with the greater common divisor of the two numbers.
    """
    if b > a:
        a, b = b, a

    while b != 0:
        r = a % b
        a = b
        b = r
    
    return a
```

### Babylonean Method for Square Root Calculation

```python
def square_root__babylonean_method(s: int) -> float:
    """
    This function has an implementation of the classic babylonean method
    to approximate the square root of a given number.

    Args:
        s(int): Input number to calculate the square root

    Returns:
        A decimal number with the square root of the number
    """
    x_n = s / 2
    diff = 0.000001

    x_n_1 = (1 / 2) * (x_n + (s / x_n))

    while abs(x_n_1 - x_n) > diff:
        x_n = x_n_1
        x_n_1 = (1 / 2) * (x_n + (s / x_n))

    return x_n
```

### Maximum of an Array

```python
def maximum_array(array: list) -> int:
    """
    This function gets the maximum value into a given array.

    Args:
        array(list): Array of numbers

    Returns:
        An integer number with the maximum in the array. If array is empty,
        a "None" value will be returned.
    """
    if len(array) > 0:
        max = array[0]
        for number in array:
            if number > max:
                max = number
    else:
        # array is empty
        max = None
    
    return max
```

# Greedy Algorithms

```python
def greedy__coin_change__iterative(denominations: list, amount: int) -> list:
    result = []
    denominations = sorted(denominations, reverse=True)
    for d in denominations:
        while amount >= d:
            result.append(d)
            amount = amount - d
    return result
```

```python
greedy__coin_change__iterative([10, 25, 1, 5], 67)
```

```python
def greedy_coin_change__maths(denominations: list, amount: int) -> list:
    result = []
    denominations = sorted(denominations, reverse=True)
    for d in denominations:
        result.append((d, amount // d))
        amount %= d
    return result
```

```python
greedy_coin_change__maths([1, 25, 10, 5], 67)
```

## Brute Force - Crack Password

```python
def get_combinations(current_size: int, max_size: int, current_string: str, alphabet: list):
    if current_size == max_size:
        return [current_string]
    else:
        result = []
        for character in alphabet:
            combination = get_combinations(current_size+1, max_size, current_string+character, alphabet)
            result.extend(combination)
        return result
    
alphabet = list("abcdefghijklmnopqrstuvwxyz._")
print(get_combinations(0, 3, '', alphabet))
```

```python
def hash(s: str) -> str:
    """Here you can change your hash function for anything you need.
    
    Args:
        s(str): String to be encrypted

    Returns:
        A string with the encrypted data
    """
    return s

def get_combinations(current_size: int, max_size: int, current_string: str, alphabet: list):
    """This function gets all the possible words based on an alphabet and a size.
    
    Args:

    """
    if current_size == max_size:
        return [current_string]
    else:
        result = []
        for character in alphabet:
            combination = get_combinations(current_size+1, max_size, current_string+character, alphabet)
            result.extend(combination)
        return result

def brute_force_password(password_original: str, min_lenght: int, max_lenght: int) -> str:
    """This function makes a brute force algorithm to find the password based on 
    encrypted password and a validation of all the posibilities.

    Args:
        password_original(str): Encrypted password to be found
        min_lenght(int): Minimum password lenght
        max_lenght(int): Maximum password lenght

    Returns:
        The password unencrypted if it exists.
    """
    alphabet = list("abcdefghijklmnopqrstuvwxyz._")

    for lenght in range(min_lenght, max_lenght + 1):
        combinations = get_combinations(0, lenght, '', alphabet)
        for candidate in combinations:
            if hash(candidate) == password_original:
                return candidate
    
    return None

brute_force_password('z.aaa.', 5, 6)
```

## N-Queens Problem

```python
def is_safe(board: list, row: int, col: int, n: int) -> bool:
    answer = True 

    # no queens in the same column
    for i in range(row):
        if board[i][col] == 'Q': 
            # there is a queen
            answer = False
            break

    # no queens in the same diagonal
    for i in range(row):
        n_r = row - i - 1
        n_c = col - i - 1
        if (n_r >= 0 and n_c >= 0) and (board[n_r][n_c] == 'Q'):
            answer = False
            break

        n_c = col + i + 1
        if (n_r >= 0 and n_c < n) and (board[n_r][n_c] == 'Q'):
            answer = False
            break
        
    return answer


def solve_n_queens(board: list, row: int, n: int):
    if row == n:
        return board
    for c in range(n):
        if is_safe(board, row, c, n):
            board[row][c] = 'Q'
            solution = solve_n_queens(board, row + 1, n)
            if solution is not None:
                return solution
            board[row][c] = ' ' # backtracking
    return None

n = 8
board = [[' ' for _ in range(n)] for __ in range(n)]
solution = solve_n_queens(board, 0, n)
for row in solution:
    print(row)
```

```python
def linear_search(a: list, x: int) -> int:
    index = -1
    for i, element in enumerate(a):
        if element == x:
            index = i
            break
    return index
```

## Bynary Search


```python
def binary_search(a: list, x: int):
    low = 0
    high = len(a) - 1
    while low <= high:
        mid = (low + high) // 2
        if a[mid] == x:
            return mid
        elif a[mid] < x:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

```python
def binary_search_recursive(a: list, x: int):
    if a == []:
        return False
    low = 0
    high = len(a) - 1
    while low <= high:
        mid = (low + high) // 2
        if a[mid] == x:
            return True
        elif a[mid] < x:
            return binary_search_recursive(a[mid + 1:], x)
        else:
            return binary_search_recursive(a[:mid], x)
    return False
```

```python
def interpolation_search(a: list, x: int):
    low = 0
    high = len(a) - 1
    # first change
    while low <= high and a[low] <= x <= a[high]:
        # second change
        mid = low + int(((x - a[low])/(a[high] - a[low])) * (high - low))
        
        if a[mid] == x:
            return mid
        elif a[mid] < x:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

```python
def books_binary_search(books: list, isbn: int):
    if books == []:
        return 'No existe el libro'
    low = 0
    high = len(books) - 1
    while low <= high:
        mid = (low + high) // 2
        if books[mid]['isbn'] == isbn:
            return books[mid]
        elif books[mid]['isbn'] < isbn:
            return books_binary_search(books[mid + 1:], isbn)
        else:
            return books_binary_search(books[:mid], isbn)
    return 'No existe el libro'
```

# SORT ALGORITHMS

## BubbleSort


```python
from typing import List
def bubble_sort(a: List) -> List:
    n = len(a)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                swap = a[j]
                a[j] = a[j + 1]
                a[j + 1] = swap
    return a
```

```python
from random import randint
a = [randint(1,100000) for _ in range(10000)]
print(a)
a = bubble_sort(a)
print(a)
```

## Selection Sort

```python
def selection_sort(a: List) -> List:
    n = len(a)
    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            if a[j] < a[min_index]:
                min_index = j
        
        swap = a[i]
        a[i] = a[min_index]
        a[min_index] = swap
    return a
```

```python
from random import randint
a = [randint(1,100000) for _ in range(10000)]
print(a)
a = selection_sort(a)
print(a)
```

```python
def selection_sort(a: List) -> List:
    n = len(a)
    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            if a[j][1] > a[min_index][1]:
                min_index = j
        
        swap = a[i]
        a[i] = a[min_index]
        a[min_index] = swap
    return a

students = [('Alice', 3.8), ("Bob", 2.5), ('Charlie', 3.9), ('Diane', 2.1), ("Eve", 3.7)]
print(selection_sort(students))
```

## N=Insertion Sort


```python
def insertion_sort(a: List) -> List:
    n = len(a)
    for i in range(2, n):
        key = a[i] 
        j = i - 1
        while j >= 1 and a[j] > key:
           a[j + 1] = a[j]
           j -= 1
        a[j + 1] = key 
```

## QuickSort

```python
from typing import List
from random import randint
def partition(a: List) -> tuple:
    index_pivot = randint(0, len(a) - 1)
    pivot = a[index_pivot]
    left = []
    right = []
    for i in range( len(a)):
        if i != index_pivot:
            if a[i] > pivot:
                right.append(a[i])
            else:
                left.append(a[i])
    return [pivot], left, right            
```

```python
def quicksort(a: List)-> List:
    if len(a) == 0:
        return []
    pivot, left, right = partition(a)
    left = quicksort(left)
    right = quicksort(right)
    return left + pivot + right
```

```python
from random import randint
a = [randint(1,100000) for _ in range(10000)]
print(a)
a = quicksort(a)
print(a)
```

## MergeSort

```python
def merge(left: List, right: List) -> List:
    index_left = 0
    index_right = 0
    result = []
    lenght_left = len(left)
    lenght_right = len(right)
    while index_left < lenght_left and \
            index_right < lenght_right:
        if left[index_left] <= right[index_right]:
          result.append(left[index_left])
          index_left += 1
        else:
          result.append(right[index_right])
          index_right += 1
    if index_left < lenght_left:
       result.extend(left[index_left:])
    else:
       result.extend(right[index_right:])
    return result
```

```python
def mergesort(a: List) -> List:
    if len(a) == 1:
        return a
    mid = len(a) // 2
    left = mergesort(a[ :mid])
    right = mergesort(a[mid: ])
    result = merge(left, right)
    return result
```

```python
from random import randint
a = [randint(1,100000) for _ in range(10000)]
print(a)
a = mergesort(a)
print(a)
```
